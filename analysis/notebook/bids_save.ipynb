```python
import numpy as np
def align_target_events(raw, psyco_data, target_events, movie_col="movie_CB.started", sound_col="sound_CB.started"):
    """
    Align target EEG events with PsychoPy onset times and update Raw annotations.

    Parameters
    ----------
    raw : mne.io.Raw
        The EEG Raw object.
    psyco_data : pd.DataFrame
        PsychoPy data containing movie and sound onset columns.
    target_events : list of str
        Event labels in raw annotations to align.
    movie_col : str
        Column name in psyco_data for movie start times.
    sound_col : str
        Column name in psyco_data for sound start times.

    Returns
    -------
    raw : mne.io.Raw
        Raw object with updated annotations for target events.
    """

    # 1. Extract events and create reverse mapping
    events, event_id = mne.events_from_annotations(raw)
    id_to_label = {v: k for k, v in event_id.items()}

    # 2. Create DataFrame with event labels and EEG times
    df_events = pd.DataFrame({
        "event_id": [id_to_label[e] for e in events[:, 2]],
        "eeg_time_sec": raw.times[events[:, 0]]
    }).drop_duplicates(subset=["event_id", "eeg_time_sec"]).reset_index(drop=True)

    # 3. Select only target events
    df_events_target = df_events[df_events["event_id"].isin(target_events)].reset_index(drop=True)

    # 4. Select PsychoPy columns and drop NaNs
    psyco_data_target = psyco_data[[movie_col, sound_col]].dropna().reset_index(drop=True)

    # 5. Align PsychoPy columns to EEG events
    df_events_target[[movie_col, sound_col]] = psyco_data_target[[movie_col, sound_col]]

    # 6. Compute onset difference and corrected EEG times
    df_events_target["onset_diff"] = df_events_target[sound_col] - df_events_target[movie_col]
    df_events_target["corrected_eeg_time_sec"] = df_events_target["eeg_time_sec"] + df_events_target["onset_diff"]

    # 7. Create new annotations for target events
    onsets = df_events_target["corrected_eeg_time_sec"].values - raw.first_samp / raw.info["sfreq"]
    durations = np.zeros_like(onsets)
    descriptions = df_events_target["event_id"].values
    new_annotations = mne.Annotations(onset=onsets, duration=durations, description=descriptions)

    # 8. Filter out old target events
    filtered_annotations = [ann for ann in raw.annotations if ann['description'] not in target_events]
    filtered_annotations = mne.Annotations(
        onset=[ann['onset'] for ann in filtered_annotations],
        duration=[ann['duration'] for ann in filtered_annotations],
        description=[ann['description'] for ann in filtered_annotations]
    )

    # 9. Combine filtered annotations with new aligned ones
    combined_annotations = filtered_annotations + new_annotations

    # 10. Set updated annotations back to raw
    raw.set_annotations(combined_annotations)

    return raw

```

```python
 # read the data
import pandas as pd


import mne
import os

import ast
import sys
from mne_bids import BIDSPath, write_raw_bids
%matplotlib qt

# Paths
root_dir = r'D:\Work_data\Notification_evoked_response\Data_at_IITD/'       # Directory with your original folders like ET001_SD043
bids_root = r'D:\Work_data\Notification_evoked_response\Data_at_IITD/bids_data'     # Where BIDS output will go
task = 'auditoryoddball'               # task name

data_type = "EEG"

# REname Channel
che_info= pd.read_excel(r'E:\IITD_One_drive-backup\researchwork\smartphone addiction\EEG_electrode_naming_issue\New_ch_name.xlsx')
dict(che_info[['Name','actual name']])

chan_rename =dict(zip(che_info['Name'], che_info['actual name']))
chan_rename = dict(list(chan_rename.items())[:-2])
import h5py

file_path = os.path.join(root_dir ,'psycho_data',"SD061_final_exp_1.2_v1_18012024_2025-04-02_13h29.35.887.hdf5")

with h5py.File(file_path, "r") as f:
    # List all groups and datasets
    def print_structure(name, obj):
        print(name)
    f.visititems(print_structure)
    
import pandas as pd


events_data = []

def extract_events(group, prefix=""):
    """Recursively extract events with time and name if available"""
    for key, item in group.items():
        if isinstance(item, h5py.Dataset):
            # Convert to numpy
            data = item[:]
            
            # Look for possible time + event columns
            cols = list(item.attrs.get("columns", [])) if "columns" in item.attrs else []
            
            if data.ndim == 2 and cols:  # looks like tabular
                df = pd.DataFrame(data, columns=[c.decode() if isinstance(c, bytes) else c for c in cols])
                
                # Try to find time + event columns
                time_col = next((c for c in df.columns if "time" in c.lower()), None)
                event_col = next((c for c in df.columns if "event" in c.lower() or "message" in c.lower()), None)
                
                if time_col and event_col:
                    for t, e in zip(df[time_col], df[event_col]):
                        events_data.append((prefix + "/" + key, t, e))
        else:
            # Recurse into subgroup
            extract_events(item, prefix + "/" + key)

with h5py.File(file_path, "r") as f:
    extract_events(f["data_collection/events"])

# Convert all extracted events into DataFrame
df_events = pd.DataFrame(events_data, columns=["Source", "Time", "Event"])
print(df_events.head())
with h5py.File(file_path, "r") as f:
    dset = f["data_collection/events/experiment/LogEvent"]
    print("Shape:", dset.shape)
    print("Dtype:", dset.dtype)
    print("Attrs:", list(dset.attrs.keys()))
    print("First row:", dset[0])

with h5py.File(file_path, "r") as f:
    dset = f["data_collection/events/experiment/MessageEvent"]

    # Basic info
    print("Shape:", dset.shape)
    print("Dtype:", dset.dtype)

    # Attributes (may contain column names or descriptions)
    print("Attributes:", dict(dset.attrs))

    # Peek at the first few rows
    try:
        for i in range(min(5, len(dset))):
            print(f"Row {i}:", dset[i])
    except Exception as e:
        print("Error reading rows:", e)
with h5py.File(file_path, "r") as f:
    dset = f["data_collection/events/experiment/MessageEvent"]

    # Basic info
    print("Shape:", dset)
    print("Dtype:", dset)

    # Attributes (often include column names, units, etc.)
    print("Attributes:", dict(dset.attrs))

    # Peek at the first few rows
    try:
        for i in range(min(1000, len(dset))):
            print(f"Row {i}:", dset[i])
    except Exception as e:
        print("Error reading rows:", e)
```
