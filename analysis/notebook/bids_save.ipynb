### Create BIDS format data


```python
data_folder = r"D:\Work_data\Notification_evoked_response\Data_at_IITD/"
subjects = [sub for sub in os.listdir(data_folder) if (("SD" in sub) & (os.path.isdir(os.path.join(data_folder,sub))))]
data_type = "EEG"



# Paths
root_dir = r'D:\Work_data\Notification_evoked_response\Data_at_IITD/'       # Directory with your original folders like ET001_SD043
bids_root = r'D:\Work_data\Notification_evoked_response\Data_at_IITD/bids_data'     # Where BIDS output will go
task = 'auditoryoddball'               # task name

# Get all subject folders and sort for consistent numbering
subject_folders = sorted([f for f in os.listdir(root_dir) if os.path.isdir(os.path.join(root_dir, f))])




for sub in subjects:
    path = os.path.join(data_folder,sub,data_type) 
    print(path)    
    for eeg in os.listdir(path):
        print(eeg)
        if eeg[-5:] == ".vhdr":
           
            raw = mne.io.read_raw_brainvision(os.path.join(path,eeg), preload=True)
            raw.rename_channels(chan_rename )
            raw = mne.add_reference_channels(raw, ref_channels=["Cz"])
            raw.set_montage("easycap-M1")
            events,event_dict = mne.events_from_annotations(raw)
            #raw.plot(scalings=dict(eeg=40e-6),clipping='transparent',duration=30.0, start=690.0, n_channels=63,color = "Blue",event_color ='black',show_options=True )
            # Generate BIDS-style subject ID: sub01, sub02, ...
            subj_id = sub.lower()  # 2-digit numbering
            print(subj_id)
            print(subj_id)
            bids_subj_id = f"{subj_id}"
            # Create BIDS path
            bids_path = BIDSPath(subject=bids_subj_id, task=task, root=bids_root, datatype='eeg',
                         suffix='eeg', extension='.vhdr')
            # Write to BIDS
            write_raw_bids(raw, bids_path=bids_path,events=events,event_id= event_dict, overwrite=True,allow_preload=True, format='BrainVision')
            print(f"Converted {data_folder} → sub-{bids_subj_id}") # it is useful to scroll eeg data 
```

```python


# Function to create participant ID
def make_participant_id(row):

    name_part = str(row['Name'])[:2].lower()
    
    gender_part = str(row['Gender'])[0].lower()
    num_part = str(row['Phone number'])[:2].lower()
    education_part = str(row['Education'])[:2].lower()
    return f"{name_part}{gender_part}{education_part}{num_part}"

# Apply the function to create a new column
exp0_df['Subject ID'] = exp0_df.apply(make_participant_id, axis=1)

exp0_df.head()
```
